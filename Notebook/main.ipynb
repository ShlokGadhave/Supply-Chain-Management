```python
# Number of Unique Values in Each Column
print("Unique Orders:", df["Order Id"].nunique())
print("Unique Customers:", df["Customer Id"].nunique())
print("Unique Products:", df["Product Name"].nunique())
print("Unique Categories:", df["Category Name"].nunique())
print("Unique Countries:", df["Customer Country"].nunique())
print("Unique Markets:", df["Market"].nunique())
```

```python
late_orders = df.loc[
    df['Order Status'] == 'PENDING','Order Id'].nunique()
total_orders = df['Order Id'].nunique()
late_percentage = (late_orders / total_orders) * 100
print(f"Late Orders: {late_orders}")
print(f"Total Orders: {total_orders}")
print(f"Late Order Percentage: {late_percentage:.2f}%")
```

```python
# pending orders 
completed_orders=df.loc[df['Order Status']=='COMPLETE','Order Id'].nunique()
total_orders=df['Order Id'].nunique()
completed_percentage=(completed_orders/total_orders)*100
print(f"Completed Orders: {completed_orders}")  
print(f'Total Orders:{total_orders}')
print(f'Complete Percentage :{completed_percentage:.2f}%')
```

```python
# Cancel Orders 
cancelled_orders=df.loc[df['Order Status']=='CANCELED','Order Id'].nunique()
total_orders=df['Order Id'].nunique()
cancelled_percentage=(cancelled_orders/total_orders)*100
print(f"Cancelled Orders: {cancelled_orders}")  
print(f'Total Orders:{total_orders}')
print(f'Cancelled Percentage :{cancelled_percentage:.2f}%')
```

```python
df.rename(columns={'order date (DateOrders)': 'Order Date'}, inplace=True)
df['Order Date']=pd.to_datetime(df['Order Date'])
df['Year']=df['Order Date'].dt.year
df['Month']=df['Order Date'].dt.month
df['Day']=df['Order Date'].dt.day
df['Month Name']=df['Order Date'].dt.month_name()
print(df[['Day','Month','Year','Month Name']].head(10))
```

Sales Analysis 

```python
# total sales generated
total_sales=df['Sales'].sum()
print(f'Total Sales Generated: ${total_sales:,.2f}')
```

```python
# total profit
profit=df['Benefit per order'].sum()
print(f'Total Profit Generated: ${profit:,.2f}')
```

```python
# average order value
avg_order_value=df['Order Profit Per Order'].mean()
print(f'Average Order Value: ${avg_order_value:,.2f}')
```

```python
# average profit per order
avg_profit_per_order=df['Order Profit Per Order'].mean()
print(f'Average Profit Per Order: ${avg_profit_per_order:,.2f}')
```

```python
# sales change over time
sales_over_time=pd.pivot_table(df,index='Year',values='Sales',aggfunc='sum')
sales_over_time
```

```python
plt.figure(figsize=(10,6))
sns.barplot(data=sales_over_time,x=sales_over_time.index,y='Sales',hue=sales_over_time.index,palette='muted')
plt.title('Disturibution of Sales Over Time')
plt.xlabel('Year')
plt.ylabel('Total Number of Sales')
```

```python
# months generate the highest sales
month_sales=df.groupby('Month Name')['Sales'].sum().sort_values(ascending=False).reset_index().rename(columns={"Sales":"Total Sales Generated by Month"})
month_sales.head(10)
```

```python
plt.Figure(figsize=(10,6))
sns.lineplot(data=month_sales,x='Total Sales Generated by Month',y='Month Name',marker='o')
plt.title('Monthly Distribution of Sales')
plt.xlabel('Month')
plt.ylabel('Total Sales')

# January has the highest sales, followed by December and March. This indicates that the company experiences a peak in sales during the winter months, possibly due to holiday shopping and seasonal demand.
```

```python
# sales disturbution
sns.histplot(data=df,x='Sales',bins=40,kde=True,color="#8e7755",edgecolor='black',alpha=0.6)
plt.title('Sales Disturbution')
```

```python
# market generates the most sales
df.columns
df["Market_Total_Sales"] = (
    df.groupby("Market")["Sales"]
      .transform("sum")
)

top_market = (
    df[["Market", "Market_Total_Sales"]]
    .drop_duplicates()
    .sort_values("Market_Total_Sales", ascending=False)
    .head(5)
)

top_market
```

```python
# country generates the most sales
df.groupby('Customer Country')['Sales'].sum().sort_values(ascending=False).reset_index(name='Total Sales')
```

Profitability Analysis

```python
# product categories generate the highest profit

profit_on_category=pd.pivot_table(data=df,index='Category Name',values='Order Profit Per Order',aggfunc='mean').sort_values(by='Order Profit Per Order',ascending=False).rename(columns={'Order Profit Per Order':'Profit'})
profit_on_category.head(10)
```

```python
# products generate the highest profit
df.columns
product_profit=df.groupby('Product Name')['Order Profit Per Order'].mean().sort_values(ascending=False).reset_index(name='Profit')
product_profit.head(10)

# Dell Laptop is the product which generates highest profit 
```

```python
# products generate negative profit
negative_profit=product_profit.sort_values(by='Profit',ascending=True)
negative_profit.head(3)
```
